#### Importer la bibliothèque

```python
import win32com.client
```

##### **1. Préparer le moteur vocal Windows**

```python
moteur = win32com.client.Dispatch("SAPI.SpVoice")
```

##### **2. Accéder aux voix OneCore installées**

```python
categorie = win32com.client.Dispatch("SAPI.SpObjectTokenCategory")

categorie.SetID(
    r"HKEY_LOCAL_MACHINE\SOFTWARE\Microsoft\Speech_OneCore\Voices",
    False
)
```

##### **3. Récupérer toutes les voix, sans filtrer les langues**

```python
voix_disponibles = []

for voix in categorie.EnumerateTokens():
    voix_disponibles.append(voix)

# Arrêter le programme si aucune voix n'est disponible
if not voix_disponibles:
    raise SystemExit("Aucune voix disponible sur cet ordinateur.")
```

##### **4. Afficher les voix et leurs numéros**

```python
print("Voici la liste des voix disponibles :")

for numero, voix in enumerate(voix_disponibles):
    nom = voix.GetDescription().removeprefix("Microsoft ")
    print(numero, ":", nom)
```

##### **5. Demander un numéro de voix valide**

```python
while True:
    try:
        numero_voix = int(
            input("\nEntrez le numéro de voix qui vous intéresse : ")
        )

        if 0 <= numero_voix < len(voix_disponibles):
            break

        print("Numéro invalide. Choisissez un numéro dans la liste.")

    except ValueError:
        print("Veuillez saisir un nombre entier.")
```

##### **6. Demander un texte non vide**

```python
texte_a_lire = input("\nEntrez le texte à lire : ").strip()

while not texte_a_lire:
    texte_a_lire = input("Veuillez entrer un texte non vide : ").strip()
```

##### **7. Choisir la voix et appliquer les paramètres**

```python
moteur.Voice = voix_disponibles[numero_voix]
moteur.Rate = -1
moteur.Volume = 100

nom_voix = voix_disponibles[numero_voix].GetDescription()
nom_voix = nom_voix.removeprefix("Microsoft ")

print("\nVoix choisie :", nom_voix)
```

##### **8. Lire le texte**

```python
moteur.Speak(texte_a_lire)

print("Lecture terminée.")
```
